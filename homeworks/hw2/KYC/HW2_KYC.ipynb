```python
from typing import List, Optional
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from pydantic import BaseModel,Field

class WorkExperience(BaseModel):
    company: str = Field(description="company name")
    role: str = Field(description="role name")
    start_year: str
    end_year: Optional[str] = Field(description="end year，if currently employed please fill in 'Present'")

class Education(BaseModel):
    school: str
    degree: str = Field(description="degree，such as BSc, MSc, PhD")

class CVData(BaseModel):
    """The key information extraction model for resumes"""
    full_name: str
    professional_title: str
    location: str
    current_company: Optional[str]
    email: Optional[str]
    education: List[Education]
    work_experience: List[WorkExperience]
    skills: List[str]
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    google_api_key=GEMINI_VERTEX_API_KEY,
    vertexai=True,
    temperature=0,
)

structured_llm = llm.with_structured_output(CVData)

prompt_template = ChatPromptTemplate.from_template(
    """
    You are a professional resume analysis assistant. Please extract the key information from the following resume text.
If certain information is missing (such as no email address), please leave it blank.
The original text may contain PDF encoding errors (such as '(cid:209)' or '(cid:123)' etc.), which usually represent icons or separators.
Please extract all the location information before and after the encoding symbols, regard these encoding errors as comma

    CV:
    {text}
    """
)

extraction_chain = prompt_template | structured_llm
```

#Matching and scoring


1. Linkedin ID search chain

```python
class li_MatchedIDs(BaseModel):
    linkedin_id: int


li_match_llm = llm.with_structured_output(li_MatchedIDs)
li_match_prompt = ChatPromptTemplate.from_template("""
    You are an identity verification expert. Please use the resume information to identify the
    corresponding person from the search results based on headline and location.

    resume information: {cv}
    LinkedIn search result: {li_res}
    """)

li_search_chain = li_match_prompt | li_match_llm
```

2. Facebook ID Match chain

```python
class fb_MatchedIDs(BaseModel):
    facebook_id: int

fb_match_llm = llm.with_structured_output(fb_MatchedIDs)
fb_match_prompt = ChatPromptTemplate.from_template("""
    You are an identity verification expert. Please use the resume information to determine whether the person found
    from the search results match the resume based on name, job, location, and education.
    Notice: 1.display name may be different, so please focus on the original name.
        2. The information may contain a discrepancy. If the matching rate is
        above 50%, please return the id; otherwise, please return 0.


    resume information: {cv}
    Facebook search result: {fb_res}
    """)
fb_search_chain = fb_match_prompt | fb_match_llm
```

3. Score chain

```python
class FinalVerdict(BaseModel):
    score: float = Field( description="The final score ranges from 0.0 to 1.0." )
    report: str = Field

judge_llm = llm.with_structured_output(FinalVerdict)

judge_prompt = ChatPromptTemplate.from_template("""
   You are an extremely meticulous expert in background check (KYC) auditing.
   Your task is to compare the claimed content in the candidate's resume
   with the actual data obtained from social media, and finally produce a
   verification report and a trust score (ranging from 0.0 to 1.0).

    [Core Data Source]
    1. resume information: {cv}
    2. LinkedIn profile: {li_prof}
    3. LinkedIn interactions: {li_interact}
    4. Facebook profile: {fb_prof}

    [Scoring standards]
    The maximum score is 1.0. Please rate based on the following three aspects comprehensively:

    Dimension 1: Core Occupation and Educational Experience (Weight 60%)
    - Carefully compare the company names, job titles, employment duration, and schools in the CV and LinkedIn profile.
    - Tolerance for Time & Status Discrepancies: Understand that CVs are static documents. If a CV states a job is
     "Current/Present" but the LinkedIn profile shows they recently left ("is_current": false) or their status is
     "open_to_work"/"hiring", THIS IS NORMAL for active job seekers. Do NOT treat this as fraud. Minor time discrepancies
     (e.g., a few months difference in start/end dates) are also fully acceptable and should not be heavily penalized.
    - Tolerance for Job Titles: Minor semantic variations in job titles are acceptable (e.g."Developer" vs. "Software Engineer").
    - If the core work experience claimed in the resume is completely absent on LinkedIn, or if there is significant
     job fraud (for example, the resume states "Senior Manager" while LinkedIn shows "Intern"), it indicates a serious
     fraud in the resume. In this case, give this part a score of 0.


    Dimension 2: Account Authenticity and Social Endorsement (Weight: 20%)
    - Check the data on the activity level of interactions on LinkedIn.
    - If the post_count is 0 and the engagement_score is extremely low, this might indicate that this is a "zombie account"
     that was created specifically for creating fake resumes. In this case, deduct the score from this dimension.
    - If there are normal posting and like-up interactions, get the full score.

    Dimension 3: Auxiliary Verification (Weight 20%)
    - Verify whether the information in the Facebook profile conflicts with the resume.
    - If Facebook fails to find this person (the data is empty), please give 0 points.
    - display name may be different, so please focus on the original name. If only display name is different,
     please do not deduct the score.

    [Outcome]
    Please, after careful consideration, output a structured JSON format, including score
    (ranging from 0.0 to 1.0) and report.
    """)

score_chain = judge_prompt | judge_llm
```

#Final Loop

```python
li_id=[]
fb_id=[]
scores=[]
score_report={}
for file_name, cv_data in structured_cvs.items():
  candidate_name = cv_data["full_name"]
  li_search_results = await tools[3].ainvoke({'q': candidate_name, 'limit': 10})
  li_matched_ids = await li_search_chain.ainvoke({
    "cv": cv_data,
    "li_res": li_search_results,
    })
  li_id.append(li_matched_ids.linkedin_id)

  fb_search_results= await tools[1].ainvoke({"user_id": li_matched_ids.linkedin_id })
  fb_matched_ids = await fb_search_chain.ainvoke({
    "cv": cv_data,
    "fb_res": fb_search_results,
  })
  fb_id.append(fb_matched_ids.facebook_id)
  linkedin_prof = await tools[4].ainvoke({'person_id': li_matched_ids.linkedin_id})
  linkedin_interact = await tools[5].ainvoke({'person_id': li_matched_ids.linkedin_id})
  grade_result = await score_chain.ainvoke({
    "cv": cv_data,
    "li_prof": linkedin_prof,
    "li_interact": linkedin_interact,
    "fb_prof": fb_search_results,
  })

  scores.append(grade_result.score)
  score_report[file_name] = grade_result.report
```

# Evaluation code

In the test phase, you will be given 5 CV files with fixed names:

    CV_1.pdf, CV_2.pdf, CV_3.pdf, CV_4.pdf, CV_5.pdf

Your system must process these CVs and output a list of 5 scores,
one score per CV, in the same order:

    scores = [s1, s2, s3, s4, s5]

Each score must be a float in the range [0, 1], representing the
reliability or confidence that the CV is valid (or meets the task criteria).

The ground-truth labels are binary:

    groundtruth = [0 or 1, ..., 0 or 1]

Each CV is evaluated independently using a threshold of 0.5:

- If score > 0.5 and groundtruth == 1 → Full credit
- If score ≤ 0.5 and groundtruth == 0 → Full credit
- Otherwise → No credit

In other words, 0.5 is the decision threshold.

- Each CV contributes equally.
- Final score = (number of correct decisions) / 5


```python
# =====================================================
#  Evaluation code
# =====================================================

def evaluate(scores, groundtruth, threshold=0.5):
    """
    scores: list of floats in [0, 1], length = 5
    groundtruth: list of ints (0 or 1), length = 5
    """
    assert len(scores) == 5
    assert len(groundtruth) == 5

    correct = 0
    decisions = []

    for s, gt in zip(scores, groundtruth):
        pred = 1 if s > threshold else 0
        decisions.append(pred)
        if pred == gt:
            correct += 1

    final_score = correct / len(scores)

    return {
        "decisions": decisions,
        "correct": correct,
        "total": len(scores),
        "final_score": final_score
    }
```
